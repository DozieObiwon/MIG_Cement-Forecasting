```python
import sqlite3
import pandas as pd

```

```python
db_path = "../data/raw/MIG_Cement_Records.db" #

conn = sqlite3.connect(db_path)
cursor = conn.cursor()
```

```python
tables = cursor.execute("SELECT name FROM sqlite_master WHERE type='table';").fetchall()
```

```python
print(tables)
```

```python
for table_name, in tables:
    print(f"****{table_name}****")

    schema = cursor.execute(f"PRAGMA table_info({table_name});").fetchall()
    display(schema)
```

```python
query = """
SELECT 
o.date, o.site_id, s.region, s.behavior, o.cement_type, o.planned_pour_tonnes, o.consumed_tonnes, o.opening_inventory_tonnes, o.deliveries_tonnes, o.closing_inventory_tonnes, o.rain_mm, o.avg_temp_c, o.silo_capacity
FROM
operations o
JOIN sites s ON o.site_id = s.site_id
"""
```

```python
df = pd.read_sql_query(query, conn)
```
